```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
import matplotlib as mpl
```

```python
mpl.rcParams['lines.linewidth'] = 2
mpl.rcParams['lines.markersize'] = 6
mpl.rcParams['figure.autolayout'] = True
mpl.rcParams['font.size'] = 15
mpl.rcParams['errorbar.capsize'] = 3
mpl.rcParams['lines.markeredgewidth'] = 1
mpl.rcParams['axes.axisbelow'] = True
```

```python
[*[1,2],3]

# --- Parámetros ---
n_muones = 100
det_size = np.array([4, 4, 0.5])
det_xy = [0,0]

def single_muon_direction():
    while True:
        theta = np.random.uniform(0,np.pi/2)
        if np.random.uniform(0,1) < np.cos(theta)**2:
            break
    phi = np.random.uniform(0,2*np.pi)
    return np.array([np.sin(theta)*np.cos(phi),np.sin(theta)*np.sin(phi),-np.cos(theta)])

def intersects(center, size, p0, p1):
    mins = center - size / 2
    maxs = center + size / 2
    d = p1 - p0
    with np.errstate(divide='ignore', invalid='ignore'):
        t1 = (mins - p0) / d
        t2 = (maxs - p0) / d
    tmin = np.max(np.minimum(t1, t2))
    tmax = np.min(np.maximum(t1, t2))
    return tmax >= max(tmin, 0)

def intersection_point(det_center, det_size, p0, p1):
    z_plane = det_center[2]
    d = p1 - p0
    if d[2] == 0:
        return None
    t = (z_plane - p0[2]) / d[2]
    if not (0 <= t <= 1):
        return None
    point = p0 + t * d
    dx, dy = det_size / 2, det_size / 2
    if abs(point[0] - det_center[0]) <= dx and abs(point[1] - det_center[1]) <= dy:
        return point
    return None

def sim_2_fijos(n_muones,det_size,det_xy,det_delta,origin_size,z0,muon_path_length):
    det1_center = np.array([*det_xy,det_delta/2])
    det2_center = np.array([*det_xy,-det_delta/2])
    
    muon_paths = []
    intersections = []
    empty_muon_paths = []
    
    for _ in range(n_muones):
        origin = np.random.uniform(-origin_size/2,origin_size/2,size=2)
        start = np.array([origin[0],origin[1],z0])
        dir = single_muon_direction()
        end = start + muon_path_length*dir
    
        if intersects(det1_center, det_size, start, end) and intersects(det2_center, det_size, start, end):
            p1 = intersection_point(det1_center, det_size, start, end)
            p2 = intersection_point(det2_center, det_size, start, end)
            if p1 is not None and p2 is not None:
                muon_paths.append((start, end))
                intersections.append((p1, p2))
        else:
            empty_muon_paths.append((start,end))
    return muon_paths,intersections,empty_muon_paths


def draw_prism(ax, center, size, color='green', alpha=0.2):
    cx, cy, cz = center
#    dx, dy, dz = size / 2
    dx = dy = dz = size/2
    v = np.array([
        [cx - dx, cy - dy, cz - dz],
        [cx + dx, cy - dy, cz - dz],
        [cx + dx, cy + dy, cz - dz],
        [cx - dx, cy + dy, cz - dz],
        [cx - dx, cy - dy, cz + dz],
        [cx + dx, cy - dy, cz + dz],
        [cx + dx, cy + dy, cz + dz],
        [cx - dx, cy + dy, cz + dz],
    ])
    faces = [
        [v[j] for j in [0, 1, 2, 3]],
        [v[j] for j in [4, 5, 6, 7]],
        [v[j] for j in [0, 1, 5, 4]],
        [v[j] for j in [2, 3, 7, 6]],
        [v[j] for j in [1, 2, 6, 5]],
        [v[j] for j in [3, 0, 4, 7]],
    ]
    ax.add_collection3d(Poly3DCollection(faces, facecolors=color, edgecolors='black', alpha=alpha))
```

```python
deltas = np.linspace(2,40,39)
detections = []
for delta in deltas:
    muon_paths,intersections,empty_muon_paths = sim_2_fijos(n_muones=100000,det_size=4,det_xy=[0,0],
                                                        det_delta=delta,origin_size=100,z0=100,
                                                        muon_path_length=400)
    detections.append(len(intersections))
detections = np.array(detections)
```

```python
fig,axs = plt.subplots(1,2,sharey=True,figsize=(12,6),layout='tight')
axs[0].scatter(deltas,detections,c='navy')
axs[1].scatter(2*np.arctan(4/deltas),detections,c='navy')
#axs[0].scatter(deltas[18],detections[18],c='crimson')
#axs[1].scatter(2*np.arctan(4/deltas[18]),detections[18],c='crimson')
fig.supylabel('Nùmero de muones detectados [u.a.]')
axs[0].set_xlabel('Distancia entre centelladores [cm]')
axs[1].set_xlabel('Ángulo de apertura [rad]')
plt.savefig('N vs dist and angle.png')
```

```python
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
draw_prism(ax, [0,0,deltas[-1]/2], 4, color='blue', alpha=0.4)
draw_prism(ax, [0,0,-deltas[-1]/2], 4, color='blue', alpha=0.4)

for (s, e), (p1, p2) in zip(muon_paths, intersections):
    ax.plot([s[0], e[0]], [s[1], e[1]], [s[2], e[2]], color='red', alpha=0.8)
    ax.scatter(*p1, color='green', s=30)
    ax.scatter(*p2, color='green', s=30)
i = 0
for (s,e) in empty_muon_paths:
    i +=1
    if i%200==0:
        ax.plot([s[0], e[0]], [s[1], e[1]], [s[2], e[2]], color='gray', alpha=0.3)

ax.set_xlim([-3, 3])
ax.set_ylim([-3, 3])
ax.set_zlim([-20/2-1, 20])
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title("Muones que atraviesan ambos detectores")
plt.tight_layout()
#ax.view_init(elev=0, azim=0)
plt.savefig('viz.png')
plt.show()
```
